```python
# Building a even odd checker.

user_input = int(input("Enter a number in which you want to check weather it is even or odd"))

if user_input % 2 == 0:
     print(f"The number you provided {user_input} is a even number.")
else:
     print(f"The number you provided {user_input} is a odd number")
```

```python
# Sample calculator.

def calculator():
     a = float(input("Enter number 1 :"))
     b = float(input("Enter number 2 :"))
     op = input("Enter operator (+,-,*,/)")

     if op == "+":
          print(f"addition of {a} + {b} =", a + b)
     elif op == "-":
          print(f"Subtraction of {a} - {b} =", a - b)
     elif op == "*":
          print(f"multiplication of {a} * {b} =", a * b)
     elif op == "/":
          if b == 0:
               print("Error divisible by zero.")
          else:
               print(f"division of {a} / {b} =", a / b)
     else:
          print("Pleae put a valid number :")

calculator()
```
